## 2.1 Dasar Teori
Pengenalan tentang operator dalam python


1. Pengertian Conditional Statement<br>
Conditional statement (pernyataan kondisional) adalah struktur kontrol dalam pemrograman yang digunakan untuk menjalankan suatu blok kode tertentu berdasarkan hasil evaluasi sebuah kondisi atau ekspresi logika (boolean). Kondisi tersebut hanya memiliki dua kemungkinan nilai, yaitu True (benar) atau False (salah). Berdasarkan nilai inilah program menentukan alur eksekusi selanjutnya.

2. Konsep Percabangan (Branching)
Struktur conditional sering disebut juga sebagai struktur percabangan, karena program tidak lagi berjalan secara linear (satu alur), melainkan bercabang menjadi beberapa kemungkinan jalur eksekusi, tergantung pada kondisi yang diperiksa. Secara umum terdapat tiga bentuk percabangan:<br>
Percabangan satu kondisi → hanya menggunakan if<br>
Percabangan dua kondisi → menggunakan if dan else<br>
Percabangan banyak kondisi → menggunakan if, elif, dan else

3. Teori Dasar Masing-Masing Pernyataan<br>
a. if<br>
Pernyataan if berfungsi mengevaluasi suatu ekspresi boolean. Apabila hasil evaluasinya bernilai True, maka blok kode yang berada di dalamnya akan dieksekusi. Sebaliknya, jika bernilai False, blok kode tersebut akan dilewati begitu saja.<br>
b. elif<br>
Pernyataan elif (kependekan dari else if) digunakan untuk memeriksa kondisi tambahan apabila kondisi pada if sebelumnya bernilai False. Dalam satu program, elif dapat digunakan lebih dari satu kali untuk menangani banyak kemungkinan kondisi secara berurutan, dan sifatnya eksklusif — artinya hanya satu blok saja yang akan dieksekusi meskipun terdapat banyak elif.<br>
c. else<br>
Pernyataan else tidak memiliki kondisi (ekspresi boolean) dan berfungsi sebagai blok "penutup" yang akan dijalankan apabila seluruh kondisi pada if dan elif di atasnya bernilai False.

4. Alur Eksekusi (Flow of Control)<br>
Secara teori, sistem akan mengevaluasi kondisi-kondisi tersebut dari atas ke bawah secara berurutan, dengan tahapan sebagai berikut:<br>
Kondisi pada if diperiksa terlebih dahulu.<br>
Jika bernilai False, sistem akan melanjutkan pemeriksaan ke elif (jika ada), satu per satu secara berurutan.<br>
Apabila seluruh kondisi pada if maupun elif bernilai False, maka blok else yang akan dijalankan.<br>
Begitu ditemukan satu kondisi yang bernilai True, blok kode terkait akan dieksekusi, dan seluruh percabangan lain yang tersisa tidak akan diperiksa lagi (diabaikan).<br>

5. Operator yang Umum Digunakan dalam Kondisi<br>
Kondisi yang diperiksa pada if maupun elif biasanya dibentuk dengan menggunakan dua jenis operator, yaitu:<br>
Operator perbandingan, seperti sama dengan, tidak sama dengan, lebih besar, lebih kecil, lebih besar atau sama dengan, dan lebih kecil atau sama dengan.<br>
Operator logika, seperti and (dan), or (atau), dan not (bukan/negasi), yang digunakan untuk menggabungkan atau membalik beberapa kondisi sekaligus
.
6. Sifat Penting Conditional Statement<br>
Bersifat eksklusif mutual, artinya hanya satu blok kode yang akan dijalankan dari keseluruhan struktur if-elif-else.<br>
Urutan pemeriksaan kondisi sangat berpengaruh terhadap hasil akhir program, sehingga penyusunan urutan kondisi perlu dilakukan secara cermat dan logis.<br>
Struktur ini menjadi dasar penting sebelum mempelajari konsep pemrograman lanjutan lainnya, seperti perulangan (looping) dan fungsi (function).

```python
# Buat program untuk kondisional if
# Program ini menggunakan fungsi input()
# Progaram menghitung diskon belanja

# input dari user
total_belanja_3011 = float(input("Masukkan Total Belanja (Rp):"))

# input status member (mengecek apakah user mengetik "y" atau "ya")
input_member_3011 = input("Apakah anda member? (y/t)").strip().lower()
is_member_3011 = input_member_3011 in ["y", "ya"]

# input status kode promo 
input_promo_3011 = input("Apakah kode promo valid? (y/t): ").strip().lower()
kode_promo_3011 = input_promo_3011 in ["y", "ya"]

total_diskon_persen_3011 = 0

# multi-if terpisah
#

if total_belanja_3011 > 1000000:
    total_diskon_persen_3011 += 10 #diskon belanja besar

if is_member_3011:
    total_diskon_persen_3011 += 5 #diskon member

if kode_promo_3011:
    total_diskon_persen_3011 += 15 #diskon promo

#menghitung nominal diskon dan total bayar
nominal_diskon_3011 = total_belanja_3011 * (total_diskon_persen_3011 / 100)
total_bayar_3011 = total_belanja_3011-nominal_diskon_3011

#output hasil
print("\n--- Rincia  Pembayaran---")
print(f"Total Diskon : {total_diskon_persen_3011}% (Rp{nominal_diskon_3011:,.0f})")
print(f"otal_bayar_3011: Rp {total_bayar_3011:,.0f}")

print(f"Total diskon yang anda dapatkan: {total_diskon_persen_3011}%")
```

```python
print ("\n=== SISTEM TRANSAKSI TOKO ===")

nama_3011 = input("Masukkan Nama Pelanggan: ")
status_3011 = input("Masukkan Status Pelanggan (member/nonmember): ")
total_belanja_3011 = int(input("Masukkan Total Belanja : "))
jumlah_barang_3011 = int(input("Masukkan Jumlah Barang : "))
promo_3011 = input("Masukkan Kode Promo : ")

 #========================================
 #  ======DATA TRANSAKSI======
 # =======================================


print("\n=== DATA TRANSAKSI ===")
print("Nama Pelanggan                     : ",nama_3011)
print("Status Pelanggan (member/nonmember): ", status_3011)
print(f"Total Belanja                      :  Rp{total_belanja_3011}")
print("Jumlah Barang                      : ", jumlah_barang_3011)
print("Kode Promo                         : ", promo_3011)

#=========================================
#=========HASIL VALIDASI================
#=========================================

kode_promo_3011 = ["HIDUP IF", "HIDUP FTI", "JALAN SEHAT"]

syarat_total_3011 = total_belanja_3011 >=200000
syarat_jumlah_barang_3011 = jumlah_barang_3011 >=3
valid_status_3011 = status_3011 == "member"
valid_promo_3011 = promo_3011 in kode_promo_3011 


print("\n=== HASIL VALIDASI ===")
print(f"Belanja >= Rp200000                : {syarat_total_3011}") 
print(f"Jumlah Barang >= 3                 : {syarat_jumlah_barang_3011}")
print(f"Status Member                      : {valid_status_3011}")
print(f"Kode Promo Tersedia                : {valid_promo_3011}")
print(f"Mendapatkan Diskon                 : {syarat_jumlah_barang_3011 or syarat_total_3011}")
print(f"Mendapatkan Promo                  : {valid_promo_3011}")

#=========================================
#  =====HASIL PERHITUNGAN============
#=========================================

print("\n=== HASIL PERHITUNGAN ===")
print(f"Diskon                             : Rp{int(0.15 * total_belanja_3011)}")
print(f"Total Pembayaran                   : Rp{int(total_belanja_3011 - 0.15 * total_belanja_3011)}")
print(f"Rata-rata Harga Barang             : Rp{int((total_belanja_3011 - 0.15 * total_belanja_3011) // jumlah_barang_3011)}")

#=========================================
#=========HAK AKSES PELANGGAN===========
#=========================================
 
print("\n=== HAK AKSES PELANGGAN ===")
print("Kode Hak Akses                     : ...")
print(f"Member Access                      : {status_3011 == "member"}")
print(f"Promo Access                       : {promo_3011 in kode_promo_3011}")
print("Free Shipping Access               : ...")

#=========================================
#=========OPERASI BITWISE===============
#=========================================


print("\n=== OPERASI BITWISE ===")

print("\n=== Kode Status Transaksi ===")

#KODE STATUS
#0001 = member
#0010 = total belanja >= Rp 200000
#0100 = jumlah barang >= 3
#1000 = promo

#menggunakan OR (|)
kode_transaksi_3011 =  int(valid_status_3011) << 0 | int(syarat_total_3011) << 1 | int(syarat_jumlah_barang_3011) << 2 | int(valid_promo_3011) << 3
kode_referensi_3011 = int(valid_status_3011) << 0 | int(syarat_total_3011) << 1 | int(valid_promo_3011) << 3

print(f"{format(int(valid_status_3011) << 0, "04b")} | {format(int(syarat_total_3011) << 1, "04b")} | {format(int(syarat_jumlah_barang_3011) << 2, "04b")} | {format(int(valid_promo_3011) << 3, "04b")}")
print(f"Kode Biner   : {format(kode_transaksi_3011, "04b")}") 
print(f"Kode Desimal : {kode_transaksi_3011}")

#====================================
#========PEMERIKSAAN STATUS=========
#====================================

print("\n=== Pemeriksaan Status ===")


print("\nCek Member")

#menggunakan and (&)

print(f"{format(kode_transaksi_3011, "04b")} & {format(int(valid_status_3011) << 0, "04b")}")
print(f"Hasil Biner   : {format((kode_transaksi_3011) & int(valid_status_3011) << 0, "04b")}")
print(f"Hasil Desimal : {(kode_transaksi_3011) & int(valid_status_3011) << 0}") 

print("\nCek Promo")

print(f"{format(kode_transaksi_3011, "04b")} & {format(int(valid_promo_3011) << 3, "04b")}")
print(f"Hasil Biner   : {format((kode_transaksi_3011) & int(valid_promo_3011) << 3, "04b")}")
print(f"Hasil Desimal  : {(kode_transaksi_3011) & int(valid_promo_3011) << 3}")

#====================================
#======PERBANDINGAN STATUS==========
#============================================

print("\n=== Perbandingan Status ===")

 # menggunakan XOR (^)

print(f"Kode Transaksi : {format(kode_transaksi_3011, "04b")}")
print(f"Kode Referensi : {format(kode_referensi_3011, "04b")}")
print(f"{format(kode_transaksi_3011, "04b")} ^ {format(kode_referensi_3011, "04b")}")
print(f"Hasil Biner    : {format((kode_transaksi_3011) ^ (kode_referensi_3011))}")
print(f"Hasil Desimal  : {(kode_transaksi_3011) ^ (kode_referensi_3011)}")

#=====================================
#=======SHIFT GESER KIRI=============
#=====================================


print("\n=== Shift ===")

#bitwise kiri/ shift kiri (<<)

print(f"{format(kode_transaksi_3011,"04b")} << 1")
print(f"Hasil Biner   : {format((kode_transaksi_3011) << 1,"04b")}")
print(f"Hasil Desimal : {(kode_transaksi_3011) << 1}") 
print("=== SELESAI ===")
```

Jika kode promo tidak terdapat dalam daftar, hasilnya adalah False.<br>

Program menggunakan operator logika or untuk menentukan apakah pelanggan mendapatkan diskon berdasarkan kondisi jumlah barang atau total belanja.<br>
Operator or akan menghasilkan True apabila salah satu atau kedua kondisi bernilai benar.<br>
Dengan demikian, berdasarkan kondisi yang dibuat dalam program, pelanggan mendapatkan status diskon apabila jumlah barang minimal 3 atau total belanja minimal Rp200.000.<br>
Perlu diperhatikan bahwa aturan ini berbeda dengan contoh awal yang menggunakan syarat member dan total belanja secara bersamaan.<br>
9. Menentukan Mendapatkan Promo<br>
Program menentukan apakah pelanggan mendapatkan promo berdasarkan validitas kode promo.<br>
Jika kode promo yang dimasukkan tersedia dalam daftar promo, pelanggan mendapatkan status True pada bagian mendapatkan promo.<br>
Jika kode promo tidak tersedia, hasilnya adalah False.<br>
10. Perhitungan Diskon<br>
Program menghitung diskon sebesar 15% dari total belanja.<br>
Perhitungan tersebut menggunakan operator aritmatika perkalian (*).<br>
Nilai 15% dikalikan dengan total belanja untuk mendapatkan jumlah potongan harga.<br>
Sebagai contoh, apabila total belanja sebesar Rp600.000, maka diskon 15% adalah Rp90.000.<br>
11. Perhitungan Total Pembayaran<br>
Setelah nilai diskon diperoleh, program menghitung total pembayaran yang harus dibayar pelanggan.<br>
Perhitungan dilakukan dengan mengurangi total belanja dengan jumlah diskon.<br>
Pada bagian ini digunakan operator aritmatika pengurangan (-).<br>
Jika total belanja Rp600.000 dan diskon Rp90.000, maka total pembayaran menjadi Rp510.000.<br>
12. Perhitungan Rata-rata Harga Barang<br>
Program juga menghitung rata-rata harga barang setelah diskon.<br>
Perhitungan dilakukan dengan membagi total pembayaran setelah diskon dengan jumlah barang yang dibeli.<br>
Pada bagian ini digunakan operator pembagian bulat (//).<br>
Operator // digunakan agar hasil pembagian berupa bilangan bulat tanpa angka desimal.<br>
Sebagai contoh, jika total pembayaran Rp510.000 dan jumlah barang 4, maka rata-rata harga barang adalah Rp56.666.<br>
13. Hak Akses Pelanggan<br>
Bagian hak akses digunakan untuk menentukan fasilitas yang dapat diperoleh pelanggan berdasarkan kondisi transaksi.<br>
Member Access digunakan untuk menunjukkan apakah pelanggan memiliki status member.<br>
Promo Access digunakan untuk menunjukkan apakah pelanggan menggunakan kode promo yang tersedia.<br>
Sementara itu, bagian Free Shipping Access pada program yang diberikan masih ditampilkan sebagai tanda titik tiga sehingga belum memiliki aturan khusus yang menentukan kapan pelanggan mendapatkan akses tersebut.<br>
14. Kode Status Transaksi<br>
Program kemudian membuat sebuah kode status transaksi menggunakan bilangan biner.<br>
Setiap kondisi transaksi diwakili oleh satu bit tertentu.<br>
Bit pertama digunakan untuk menunjukkan status member.<br>
Bit kedua digunakan untuk menunjukkan apakah total belanja memenuhi minimal Rp200.000.<br>
Bit ketiga digunakan untuk menunjukkan apakah jumlah barang minimal 3.<br>
Bit keempat digunakan untuk menunjukkan apakah kode promo tersedia.<br>
Dengan cara tersebut, beberapa kondisi transaksi dapat digabungkan menjadi satu kode status transaksi.<br>
15. Operator Bitwise OR<br>
Dalam pembentukan kode status transaksi digunakan operator bitwise OR (|).<br>
Operator OR digunakan untuk menggabungkan beberapa nilai bit menjadi satu kode transaksi.<br>
Jika semua kondisi bernilai benar, setiap bit akan bernilai 1 sehingga menghasilkan kode biner 1111.<br>
Kode biner 1111 memiliki nilai desimal 15.<br>
16. Operator Bitwise AND<br>
Program menggunakan operator bitwise AND (&) untuk melakukan pemeriksaan terhadap status tertentu dari kode transaksi.<br>
Pemeriksaan pertama digunakan untuk mengecek status member.<br>
Jika bit member aktif, hasil pemeriksaan akan menunjukkan nilai biner 0001 atau nilai desimal 1.<br>
Pemeriksaan kedua digunakan untuk mengecek status promo.<br>
Jika bit promo aktif, hasilnya adalah 1000 atau nilai desimal 8.<br>
Dengan demikian, operator AND digunakan untuk mengambil atau memeriksa bagian tertentu dari kode status transaksi.<br>
17. Operator Bitwise XOR<br>
Program juga menggunakan operator bitwise XOR (^) untuk membandingkan kode transaksi dengan kode referensi.<br>
XOR akan menghasilkan nilai 0 apabila dua bit yang dibandingkan sama dan menghasilkan nilai 1 apabila kedua bit berbeda.<br>
Hasil XOR dapat digunakan untuk mengetahui bagian kode transaksi yang memiliki perbedaan dengan kode referensi.<br>
Sebagai contoh, kode transaksi 1111 dibandingkan dengan kode referensi 1011 menghasilkan 0100.<br>
Nilai biner 0100 memiliki nilai desimal 4.<br>
18. Operator Shift Kiri<br>
Program menggunakan operator shift kiri (<<) untuk menggeser posisi bit ke sebelah kiri.<br>
Pada contoh, kode transaksi 1111 digeser satu posisi ke kiri sehingga menjadi 11110.<br>
Nilai desimal dari 1111 adalah 15, sedangkan setelah digeser satu posisi ke kiri nilainya menjadi 30.<br>
Secara sederhana, pergeseran satu posisi ke kiri memiliki efek yang sama dengan mengalikan nilai dengan 2.<br>
19. Fungsi Format Biner<br>
Program menggunakan fungsi format untuk menampilkan nilai angka dalam bentuk bilangan biner.<br>
Penggunaan format empat digit biner membuat kode status lebih mudah dibaca dan sesuai dengan konsep empat kondisi yang digunakan dalam transaksi.<br>
Contohnya, angka desimal 15 dapat ditampilkan sebagai 1111 dalam bentuk biner.<br>

Jawaban dan penjelasan dari analisi laporan dari tugas3:

1. Operator aritmatika yang digunakan<br>
Operator aritmatika yang digunakan dalam program adalah perkalian, pengurangan, dan pembagian bulat. Perkalian digunakan untuk menghitung besar diskon dari total belanja. Pengurangan digunakan untuk mendapatkan total pembayaran setelah diskon. Sedangkan pembagian bulat digunakan untuk menghitung rata-rata harga barang berdasarkan total pembayaran dan jumlah barang. Operator tersebut membantu program melakukan perhitungan transaksi secara otomatis.

2. Perbedaan operator and, or, dan not<br>
Operator and digunakan ketika semua kondisi harus terpenuhi agar menghasilkan nilai True. Operator or digunakan ketika cukup salah satu kondisi terpenuhi untuk menghasilkan True. Sedangkan operator not digunakan untuk membalik hasil suatu kondisi, yaitu True menjadi False dan False menjadi True.<br>
Pada program yang dibuat, konsep operator logika dapat digunakan untuk menentukan apakah pelanggan memenuhi beberapa persyaratan sekaligus, misalnya jumlah belanja dan jumlah barang harus memenuhi batas tertentu.

3. Fungsi operator in dan not in pada pemeriksaan kode promo<br>
Operator in digunakan untuk memeriksa apakah kode promo yang dimasukkan pelanggan terdapat dalam daftar kode promo yang tersedia. Jika kode yang dimasukkan terdapat dalam daftar, maka hasil pemeriksaannya adalah True.<br>
Sedangkan not in digunakan untuk memeriksa apakah kode promo tidak terdapat dalam daftar. Jika kode tersebut tidak tersedia, maka hasilnya adalah True.<br>
Dalam program, operator ini digunakan untuk menentukan apakah pelanggan mendapatkan akses terhadap promo yang tersedia.

4. Perbedaan is dan == pada Python<br>
Operator == digunakan untuk membandingkan nilai dari dua data. Sedangkan operator is digunakan untuk memeriksa apakah dua objek merupakan objek yang sama.<br>
Dalam program transaksi, operator == digunakan untuk membandingkan status pelanggan, misalnya untuk mengetahui apakah status pelanggan adalah member<br>
Operator is tidak digunakan karena program hanya membutuhkan perbandingan nilai, bukan pemeriksaan objek yang sama.

5. Penggunaan operator bitwise untuk menentukan hak akses pelanggan<br>
Operator bitwise digunakan untuk mengolah kode hak akses dalam bentuk bilangan biner. Setiap posisi bit dapat mewakili status atau hak akses tertentu, seperti member dan promo.<br>
Operator OR (|) digunakan untuk menggabungkan beberapa kode status menjadi satu kode transaksi. Kemudian operator AND (&) digunakan untuk memeriksa apakah suatu hak akses tertentu aktif.<br>
Selain itu, program menggunakan XOR (^) untuk membandingkan kode transaksi dengan kode referensi. Operator shift kiri (<<) digunakan untuk menggeser posisi bit ke sebelah kiri.<br>
Dengan cara tersebut, hak akses pelanggan dapat direpresentasikan menggunakan kode biner.

Setelah proses validasi selesai, program menghitung diskon dan total pembayaran. Program juga menghitung rata-rata harga barang berdasarkan total pembayaran dan jumlah barang.<br>
